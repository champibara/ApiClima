```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import levene
```

```python
API_KEY = "170265dcc664b1396cd1d6be03d32bc0"
```

```python
def obtener_temperaturas(ciudad):
    url = f"https://api.openweathermap.org/data/2.5/forecast?q={ciudad}&appid={API_KEY}&units=metric"
    response = requests.get(url)
    data = response.json()

    temperaturas = []
    fechas = []

    for item in data['list']:
        temperaturas.append(item['main']['temp'])
        fechas.append(item['dt_txt'])

    df = pd.DataFrame({
        'Fecha': fechas,
        'Temperatura': temperaturas
    })

    return df
```

```python
lima = obtener_temperaturas("Lima")
lima.head()
```

```python
trujillo = obtener_temperaturas("Trujillo")
trujillo.head()
```

```python
#Creamos un boxplot para obtener una comparación visual
plt.boxplot(
    [lima["Temperatura"], trujillo["Temperatura"]],
    labels=["Lima", "Trujillo"]
)
plt.ylabel("Temperatura (°C)")
plt.title("Comparación de temperaturas: Lima vs Trujillo")
plt.show()
```

```python
# Prueba de Levene
from scipy.stats import levene

stat, p_value = levene(
    lima["Temperatura"],
    trujillo["Temperatura"]
)

print("Estadístico de Levene:", stat)
print("p-value:", p_value)
```

### Prueba de Levene – Homogeneidad de varianzas

Se aplicó la prueba de Levene para evaluar si las varianzas de la temperatura entre Lima y Trujillo son iguales.

El estadístico de Levene obtenido fue 0.487 y el valor p fue 0.487.

Dado que el valor p es mayor a 0.05, no se rechaza la hipótesis nula, por lo que se puede asumir que las varianzas de la temperatura en ambas ciudades son homogéneas.
